```python
import re

def is_garbage(text):
    text = str(text).strip()

    # Empty
    if not text:
        return True

    # Only punctuation/symbols
    if not re.search(r'[\u0980-\u09FFa-zA-Z0-9]', text):
        return True

    # Random English/Latin characters like ikgbdtk
    if re.fullmatch(r'[a-zA-Z]{5,}', text):
        return True

    # Repeated characters like aaaaaaa / !!!!!!!
    if re.fullmatch(r'(.)\1{4,}', text):
        return True

    return False


garbage_mask = train["input"].apply(is_garbage)

train[garbage_mask][["id", "input"]]
```

```python
# ============================================
# CHECK EXTRA SPACES / SYMBOLS / PUNCTUATION
# ============================================

import re
import pandas as pd

check_df = df_clean.copy()

# Make sure input is string
check_df["input"] = check_df["input"].fillna("").astype(str)


# ============================================
# 1. LEADING / TRAILING SPACES
# ============================================

check_df["leading_space"] = check_df["input"].str.match(r"^\s")
check_df["trailing_space"] = check_df["input"].str.match(r"\s$")

print("Leading spaces:", check_df["leading_space"].sum())
print("Trailing spaces:", check_df["trailing_space"].sum())


# ============================================
# 2. MULTIPLE / EXTRA SPACES
# ============================================

check_df["multiple_spaces"] = check_df["input"].str.contains(
    r"\s{2,}",
    regex=True
)

print("Multiple consecutive spaces:", check_df["multiple_spaces"].sum())


# ============================================
# 3. PUNCTUATION / SYMBOLS
# ============================================

# Anything that is NOT:
# Bengali characters, English letters, numbers, or whitespace
check_df["has_symbols"] = check_df["input"].str.contains(
    "[^\u0980-\u09FFa-zA-Z0-9\s]",
    regex=True
)

print("Rows containing symbols/punctuation:", check_df["has_symbols"].sum())


# ============================================
# 4. SHOW ROWS WITH EXTRA SPACES
# ============================================

space_issues = check_df[
    check_df["leading_space"]
    | check_df["trailing_space"]
    | check_df["multiple_spaces"]
]

print("\nROWS WITH SPACE ISSUES")
print("----------------------")
display(
    space_issues[
        ["id", "input", "leading_space",
         "trailing_space", "multiple_spaces"]
    ]
)


# ============================================
# 5. SHOW ROWS WITH SYMBOLS / PUNCTUATION
# ============================================

symbol_issues = check_df[check_df["has_symbols"]]

print("\nROWS WITH SYMBOLS / PUNCTUATION")
print("--------------------------------")
display(
    symbol_issues[
        ["id", "input"]
    ]
)


# ============================================
# 6. SUMMARY
# ============================================

print("\nSUMMARY")
print("-------")
print("Total rows:", len(check_df))
print("Leading space:", check_df["leading_space"].sum())
print("Trailing space:", check_df["trailing_space"].sum())
print("Multiple spaces:", check_df["multiple_spaces"].sum())
print("Symbols/punctuation:", check_df["has_symbols"].sum())
```

```python
import re
import pandas as pd

# ============================================
# CLEAN TEXT
# Removes:
# - English + Bengali phone/contact numbers
# - Email addresses
# - WWW / HTTP / HTTPS URLs
# - Website/domain links
# - Markdown links
# ============================================

def clean_text(text):
    if pd.isna(text):
        return text

    text = str(text)

    # ------------------------------------------------
    # 1. Remove markdown links
    # Example:
    # [website](http://example.com)
    # [email](mailto:test@gmail.com)
    # ------------------------------------------------
    text = re.sub(
        r'\[[^\]]*\]\(\s*(?:https?://|mailto:)[^)]+\)',
        ' ',
        text,
        flags=re.IGNORECASE
    )

    # ------------------------------------------------
    # 2. Remove email addresses
    # ------------------------------------------------
    text = re.sub(
        r'\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}\b',
        ' ',
        text
    )

    # ------------------------------------------------
    # 3. Remove URLs
    # http://
    # https://
    # www.
    # ------------------------------------------------
    text = re.sub(
        r'(?i)\b(?:https?://|www\.)[^\s<>()\[\]]+',
        ' ',
        text
    )

    # ------------------------------------------------
    # 4. Remove domain-style websites
    # Example:
    # ChatDoctor.com
    # example.org
    # website.net
    # ------------------------------------------------
    text = re.sub(
        r'(?i)\b(?:[a-z0-9-]+\.)+'
        r'(?:com|org|net|info|biz|io|co|in|bd|au)'
        r'(?:/[^\s]*)?',
        ' ',
        text
    )

    # ------------------------------------------------
    # 5. Remove phone/contact numbers
    #
    # Supports BOTH:
    # English digits: 0123456789
    # Bengali digits: ০১২৩৪৫৬৭৮৯
    #
    # Also handles:
    # +880 1712 345678
    # 01712-345678
    # ০১৭১২-৩৪৫৬৭৮
    # ০১৭১২ ৩৪৫৬৭৮
    # ------------------------------------------------

    english_digits = r'0-9'
    bengali_digits = r'০-৯'

    # English + Bengali digit character class
    digits = f'{english_digits}{bengali_digits}'

    # Numbers with spaces, -, brackets, +, dots
    phone_pattern = (
        rf'(?<![{digits}])'
        rf'\+?[{digits}]'
        rf'(?:[\s().-]*[{digits}]){{7,}}'
        rf'(?![{digits}])'
    )

    text = re.sub(phone_pattern, ' ', text)

    # ------------------------------------------------
    # 6. Remove leftover mailto
    # ------------------------------------------------
    text = re.sub(
        r'(?i)\bmailto\s*:\s*',
        ' ',
        text
    )

    # ------------------------------------------------
    # 7. Normalize spaces
    # ------------------------------------------------
    text = re.sub(r'[ \t]+', ' ', text)

    # ------------------------------------------------
    # 8. Clean spaces around newlines
    # ------------------------------------------------
    text = re.sub(r' *\n *', '\n', text)

    # ------------------------------------------------
    # 9. Remove excessive blank lines
    # ------------------------------------------------
    text = re.sub(r'\n{3,}', '\n\n', text)

    return text.strip()


# ============================================
# CREATE NEW DATAFRAME
# ============================================

df_final = df_cleaned.copy()


# ============================================
# CLEAN BOTH INPUT AND OUTPUT
# ============================================

df_final["input"] = df_final["input"].apply(clean_text)
df_final["output"] = df_final["output"].apply(clean_text)


# ============================================
# CHECK SHAPE
# ============================================

print("Before cleaning :", df_cleaned.shape)
print("After cleaning  :", df_final.shape)


# ============================================
# CHECK EXAMPLES
# ============================================

display(
    df_final[["id", "input", "output"]].head(10)
)
```

```python
# 1. Load Data


# 2. Setup System Prompt & Chat Formatting
SYSTEM_PROMPT = (
    "আপনি একজন অভিজ্ঞ ও সহানুভূতিশীল চিকিৎসক। রোগীর বর্ণনা পড়ে "
    "স্পষ্ট, নির্ভুল এবং সহজবোধ্য ভাষায় চিকিৎসা পরামর্শ দিন।"
)

# 3. Model & Tokenizer Setup via Unsloth
MAX_SEQ_LENGTH = 1024
MODEL_ID = "Qwen/Qwen2.5-1.5B-Instruct"
OUTPUT_DIR = "bn-doctor-lora"

model, tokenizer = FastLanguageModel.from_pretrained(
    model_name=MODEL_ID,
    max_seq_length=MAX_SEQ_LENGTH,
    dtype=None,
    load_in_4bit=True,
)

tokenizer = get_chat_template(tokenizer, chat_template="qwen-2.5")


# 4. Format Labeled Train Dataset
def format_prompts(df_train):
    records = df_train.to_dict("records")
    formatted_texts = [
        tokenizer.apply_chat_template(
            [
                {"role": "system", "content": SYSTEM_PROMPT},
                {"role": "user", "content": row["input"]},
                {"role": "assistant", "content": row["output"]},
            ],
            tokenize=False,
        )
        for row in records
    ]
    return pd.DataFrame({"text": formatted_texts})


df_formatted = format_prompts(df_cleaned_base)
ds = Dataset.from_pandas(df_formatted)

# Validation split
split = ds.train_test_split(test_size=500, seed=42)
train_ds, eval_ds = split["train"], split["test"]

print(
    f"Dataset ready! Training samples: {len(train_ds)} | Eval samples: {len(eval_ds)}"
)
```

```python
import torch
from transformers import TrainingArguments
from trl import SFTTrainer
from unsloth import FastLanguageModel, is_bfloat16_supported
from unsloth.chat_templates import train_on_responses_only

# 0. Subsample so this actually finishes within a Kaggle session
#    (only run this line once — if you rerun the cell, it'll re-shuffle from
#    the already-subsampled 35k, which is fine, just don't chain it further)
train_ds = train_ds.shuffle(seed=42).select(range(35000))

# 1. Add LoRA Adapters
model = FastLanguageModel.get_peft_model(
    model,
    r=16,
    target_modules=[
        "q_proj", "k_proj", "v_proj", "o_proj",
        "gate_proj", "up_proj", "down_proj",
    ],
    lora_alpha=32,
    lora_dropout=0,
    bias="none",
    use_gradient_checkpointing="unsloth",
    random_state=42,
)

# 2. Setup SFTTrainer
trainer = SFTTrainer(
    model=model,
    tokenizer=tokenizer,
    train_dataset=train_ds,
    eval_dataset=eval_ds,
    dataset_text_field="text",
    max_seq_length=1024,
    dataset_num_proc=1,
    packing=False,
    args=TrainingArguments(
        per_device_train_batch_size=2,       # small enough to avoid CPU offload
        gradient_accumulation_steps=16,       # effective batch = 32
        warmup_steps=50,
        num_train_epochs=1,
        average_tokens_across_devices=False,  # fixes the 'int' object has no attribute 'mean' crash
        learning_rate=2e-4,
        fp16=not is_bfloat16_supported(),
        bf16=is_bfloat16_supported(),
        logging_steps=50,
        optim="adamw_8bit",
        weight_decay=0.01,
        lr_scheduler_type="cosine",
        seed=42,
        output_dir=OUTPUT_DIR,
        eval_strategy="steps",
        eval_steps=250,
        save_strategy="steps",
        save_steps=250,
        save_total_limit=2,
        load_best_model_at_end=True,
        metric_for_best_model="eval_loss",
        report_to="none",
    ),
)

# 3. Apply Response Masking (Train only on doctor responses)
trainer = train_on_responses_only(
    trainer,
    instruction_part="<|im_start|>user\n",
    response_part="<|im_start|>assistant\n",
    num_proc=1,
)

# 4. Argument Hotfix (separate from the average_tokens_across_devices fix above —
#    this one guards num_items_in_batch typing, doesn't hurt to keep)
_orig_training_step = trainer.training_step
def safe_training_step(*args, **kwargs):
    args_list = list(args)
    if len(args_list) >= 3 and isinstance(args_list[2], int):
        args_list[2] = torch.tensor(args_list[2], dtype=torch.float32)
    if "num_items_in_batch" in kwargs and isinstance(kwargs["num_items_in_batch"], int):
        kwargs["num_items_in_batch"] = torch.tensor(kwargs["num_items_in_batch"], dtype=torch.float32)
    return _orig_training_step(*args_list, **kwargs)
trainer.training_step = safe_training_step

# 5. Start Training Loop
print("Starting Unsloth Fine-Tuning...")
trainer.train()

# 6. Save Adapter
model.save_pretrained(f"{OUTPUT_DIR}/final_adapter")
tokenizer.save_pretrained(f"{OUTPUT_DIR}/final_adapter")
print(f"🎉 Training Complete! Adapter saved to {OUTPUT_DIR}/final_adapter")
```

```python
import torch
from peft import PeftModel
from transformers import AutoModelForCausalLM, AutoTokenizer

BASE_MODEL_ID = "Qwen/Qwen2.5-1.5B-Instruct"
ADAPTER_PATH = "bn-doctor-lora/final_adapter"

SYSTEM_PROMPT = (
    "আপনি একজন অভিজ্ঞ ও সহানুভূতিশীল চিকিৎসক। রোগীর বর্ণনা পড়ে "
    "স্পষ্ট, নির্ভুল এবং সহজবোধ্য ভাষায় চিকিৎসা পরামর্শ দিন।"
)

# 1. Load Tokenizer & Base Model
print("Loading tokenizer and base model...")
tokenizer = AutoTokenizer.from_pretrained(ADAPTER_PATH)

base_model = AutoModelForCausalLM.from_pretrained(
    BASE_MODEL_ID, torch_dtype=torch.float16, device_map="auto"
)

# 2. Load Fine-Tuned LoRA Adapter
print("Loading LoRA adapter...")
model = PeftModel.from_pretrained(base_model, ADAPTER_PATH)
model.eval()


# 3. Define Inference Function
def get_medical_advice(user_query: str) -> str:
    messages = [
        {"role": "system", "content": SYSTEM_PROMPT},
        {"role": "user", "content": user_query},
    ]

    # Format using Qwen chat template with generation prompt
    prompt = tokenizer.apply_chat_template(
        messages, tokenize=False, add_generation_prompt=True
    )

    inputs = tokenizer(prompt, return_tensors="pt").to(model.device)

    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            max_new_tokens=512,
            temperature=0.7,
            top_p=0.9,
            repetition_penalty=1.1,
            do_sample=True,
            pad_token_id=tokenizer.pad_token_id,
        )

    # Extract only newly generated text (exclude the input prompt)
    generated_tokens = outputs[0][inputs.input_ids.shape[1] :]
    return tokenizer.decode(generated_tokens, skip_special_tokens=True)


# 4. Test Example
test_patient_query = (
    "আমার গত ৩ দিন ধরে হালকা জ্বর এবং গলা ব্যথা করছে। সাথে সামান্য কাশি আছে। "
    "আমার এখন কী করা উচিত?"
)

print(f"\n--- Patient Input ---\n{test_patient_query}\n")
response = get_medical_advice(test_patient_query)
print(f"--- Doctor Response ---\n{response}")
```

```python
import pandas as pd
import torch
from datasets import Dataset
from tqdm import tqdm
from unsloth import FastLanguageModel

# 1. Reconstruct the held-out split
ds_raw = Dataset.from_pandas(df_cleaned_base[["input", "output"]])
eval_raw = ds_raw.train_test_split(test_size=500, seed=42)["test"]

# 2. Configuration & Unsloth Model Loading
ADAPTER_PATH = (
    "/kaggle/input/datasets/towshintushi/adapter/bn-doctor-lora/final_adapter"
)
SYSTEM_PROMPT = (
    "আপনি একজন অভিজ্ঞ ও সহানুভূতিশীল চিকিৎসক। রোগীর বর্ণনা পড়ে "
    "স্পষ্ট, নির্ভুল এবং সহজবোধ্য ভাষায় চিকিৎসা পরামর্শ দিন।"
)

print("Loading model and adapter via Unsloth engine...")
model, tokenizer = FastLanguageModel.from_pretrained(
    model_name=ADAPTER_PATH,
    max_seq_length=2048,
    dtype=None,
    load_in_4bit=False,
)
FastLanguageModel.for_inference(model)

# Configure left padding for batch generation
tokenizer.padding_side = "left"
if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token

eos_token_id = tokenizer.convert_tokens_to_ids("<|im_end|>")
if eos_token_id is None:
    eos_token_id = tokenizer.eos_token_id


def clean_doctor_response(text):
    if not isinstance(text, str) or not text:
        return ""
    stop_markers = [
        "<|im_end|>",
        "<|im_start|>",
        "<|endoftext|>",
        "user\n",
        "assistant\n",
    ]
    for marker in stop_markers:
        if marker in text:
            text = text.split(marker)[0]
    return text.strip()


# 3. Batch Generation
BATCH_SIZE = 16
inputs_list = eval_raw["input"]
outputs_list = eval_raw["output"]
preds = []

print(f"Generating responses for {len(inputs_list)} validation samples...")

for i in tqdm(range(0, len(inputs_list), BATCH_SIZE)):
    batch_prompts = inputs_list[i : i + BATCH_SIZE]
    texts = [
        tokenizer.apply_chat_template(
            [
                {"role": "system", "content": SYSTEM_PROMPT},
                {"role": "user", "content": p},
            ],
            tokenize=False,
            add_generation_prompt=True,
        )
        for p in batch_prompts
    ]

    enc = tokenizer(
        texts,
        return_tensors="pt",
        padding=True,
        truncation=True,
        max_length=1024,
    ).to("cuda")

    with torch.no_grad():
        out = model.generate(
            **enc,
            max_new_tokens=650,
            do_sample=False,
            repetition_penalty=1.15,
            eos_token_id=eos_token_id,
            pad_token_id=tokenizer.pad_token_id,
        )

    input_len = enc["input_ids"].shape[1]
    for j in range(len(batch_prompts)):
        raw_text = tokenizer.decode(
            out[j][input_len:], skip_special_tokens=True
        )
        preds.append(clean_doctor_response(raw_text))

# 4. Save CSV
val_predictions = pd.DataFrame({"output": outputs_list, "doctor_response": preds})
val_predictions.to_csv("val_predictions.csv", index=False)
print(f"\n🎉 Successfully saved {len(val_predictions)} rows to val_predictions.csv")
```

```python
import numpy as np
import pandas as pd
from collections import Counter
from bert_score import score as compute_bert_score
import evaluate

# ---------------------------------------------------------
# 1. Load Ground Truth vs Generated Predictions
# ---------------------------------------------------------
df_eval = pd.read_csv("val_predictions.csv")
references = df_eval["output"].tolist()
predictions = df_eval["doctor_response"].tolist()
print(f"Evaluating {len(predictions)} validation samples...\n")

# ---------------------------------------------------------
# 2. Compute BERTScore (F1, Precision, Recall)
# ---------------------------------------------------------
print("Computing BERTScore...")
P, R, F1 = compute_bert_score(
    predictions=predictions,
    references=references,
    lang="bn",
    verbose=True,
    batch_size=32,
    device="cuda",
)
bert_f1 = F1.mean().item()
bert_precision = P.mean().item()
bert_recall = R.mean().item()

# ---------------------------------------------------------
# 3. Compute ROUGE Scores
#    NOTE: must override the tokenizer — evaluate's default ROUGE
#    tokenizer strips non-ASCII characters, which silently zeroes
#    out every Bengali score without erroring.
# ---------------------------------------------------------
print("Computing ROUGE scores...")
rouge_metric = evaluate.load("rouge")
rouge_results = rouge_metric.compute(
    predictions=predictions,
    references=references,
    use_aggregator=True,
    tokenizer=lambda x: x.split(),
)

# ---------------------------------------------------------
# 4. Compute Token-level F1 (this is the 30% competition sub-metric —
#    NOT the same as BLEU or ROUGE-1, and wasn't computed before)
# ---------------------------------------------------------
print("Computing Token-level F1...")
def token_f1(pred, ref):
    p_toks, r_toks = pred.split(), ref.split()
    if not p_toks or not r_toks:
        return 0.0
    common = Counter(p_toks) & Counter(r_toks)
    num_common = sum(common.values())
    if num_common == 0:
        return 0.0
    precision = num_common / len(p_toks)
    recall = num_common / len(r_toks)
    return 2 * precision * recall / (precision + recall)

token_f1_scores = [token_f1(p, r) for p, r in zip(predictions, references)]
token_f1_mean = np.mean(token_f1_scores)

# ---------------------------------------------------------
# 5. Compute BLEU Score (informational only — not part of the
#    competition's actual scoring formula)
# ---------------------------------------------------------
print("Computing BLEU score...")
bleu_metric = evaluate.load("bleu")
bleu_results = bleu_metric.compute(
    predictions=predictions, references=references
)

# ---------------------------------------------------------
# 6. Composite Score — this is the real number to track,
#    approximates your Phase 1 leaderboard score
# ---------------------------------------------------------
composite_score = 0.5 * bert_f1 + 0.3 * token_f1_mean + 0.2 * rouge_results["rougeL"]

# ---------------------------------------------------------
# 7. Display Leaderboard Evaluation Summary
# ---------------------------------------------------------
print("=" * 50)
print("              COMPETITION METRICS              ")
print("=" * 50)
print(f"BERTScore F1        : {bert_f1:.4f}")
print(f"BERTScore Precision : {bert_precision:.4f}")
print(f"BERTScore Recall    : {bert_recall:.4f}")
print("-" * 50)
print(f"Token-level F1      : {token_f1_mean:.4f}")
print("-" * 50)
print(f"ROUGE-1             : {rouge_results['rouge1']:.4f}")
print(f"ROUGE-2             : {rouge_results['rouge2']:.4f}")
print(f"ROUGE-L             : {rouge_results['rougeL']:.4f}")
print("-" * 50)
print(f"BLEU-4 (info only)  : {bleu_results['bleu']:.4f}")
print("=" * 50)
print(f"ESTIMATED LEADERBOARD SCORE : {composite_score:.4f}")
print("=" * 50)
```

```python
import re
import pandas as pd

def clean_medical_dataset(df):
    initial_count = len(df)
    
    # 1. Remove duplicates across input and output
    df = df.drop_duplicates(subset=["input"]).copy()
    df = df.drop_duplicates(subset=["output"]).copy()
    
    # 2. Filter out overly short or empty responses
    df = df[df["input"].str.strip().str.len() > 10]
    df = df[df["output"].str.strip().str.len() > 30]
    
    # 3. Remove phrase-repetition loops within a single response
    def has_repetition_loop(text):
        # Detects if a sequence of 3+ words repeats 3 or more times
        pattern = r'(\b[\u0980-\u09FF\w\s]{5,30}\b)\1{2,}'
        return bool(re.search(pattern, text))
    
    df = df[~df["output"].apply(has_repetition_loop)]
    
    # 4. Filter out common AI fluff/filler intros that cause repetitive patterns
    fluff_patterns = [
        r"^হেলো, আপনার প্রশ্নটি পোস্ট করার জন্য ধন্যবাদ।",
        r"^আমি আপনার উদ্বেগ বুঝতে পারছি।",
    ]
    
    for pattern in fluff_patterns:
        df["output"] = df["output"].str.replace(pattern, "", regex=True).str.strip()
        
    # 5. Remove hallucinated / non-sensical terms
    # Add any specific fabricated words you observed to this list
    hallucinated_terms = ["মায়োস্ট্রোমায়েড", "MYDOSIS"]
    for term in hallucinated_terms:
        df = df[~df["output"].str.contains(term, na=False)]

    print(f"Dataset cleaned: reduced from {initial_count} to {len(df)} samples.")
    return df.reset_index(drop=True)

# Clean your dataframe before formatting for SFTTrainer
df_cleaned_base = clean_medical_dataset(df_cleaned_base)
```

```python
import os
import torch
from datasets import Dataset
from transformers import TrainingArguments
from trl import SFTTrainer
from unsloth import FastLanguageModel, is_bfloat16_supported
from unsloth.chat_templates import get_chat_template, train_on_responses_only

# 1. Model Configuration
MODEL_NAME = "Qwen/Qwen2.5-3B-Instruct"
MAX_SEQ_LENGTH = 1024  # Reduced from 2048 to 1024 for 2x faster matrix ops
OUTPUT_DIR = "bn-doctor-qwen3b-lora"

SYSTEM_PROMPT = (
    "আপনি একজন অভিজ্ঞ ও সহানুভূতিশীল চিকিৎসক। রোগীর বর্ণনা পড়ে "
    "স্পষ্ট, নির্ভুল এবং সহজবোধ্য ভাষায় চিকিৎসা পরামর্শ দিন।"
)

# 2. Load Base Model and Tokenizer
model, tokenizer = FastLanguageModel.from_pretrained(
    model_name=MODEL_NAME,
    max_seq_length=MAX_SEQ_LENGTH,
    load_in_4bit=True,
)

tokenizer = get_chat_template(
    tokenizer,
    chat_template="qwen-2.5",
)

# 3. Apply LoRA
model = FastLanguageModel.get_peft_model(
    model,
    r=16,
    lora_alpha=16,
    lora_dropout=0,
    target_modules=[
        "q_proj", "k_proj", "v_proj", "o_proj",
        "gate_proj", "up_proj", "down_proj",
    ],
    bias="none",
    use_gradient_checkpointing="unsloth",
    random_state=42,
)

# 4. Format Dataset
def format_prompts(examples):
    texts = []
    for inp, out in zip(examples["input"], examples["output"]):
        messages = [
            {"role": "system", "content": SYSTEM_PROMPT},
            {"role": "user", "content": inp},
            {"role": "assistant", "content": out},
        ]
        text = tokenizer.apply_chat_template(
            messages, tokenize=False, add_generation_prompt=False
        )
        texts.append(text)
    return {"text": texts}

dataset = Dataset.from_pandas(df_cleaned_base)
dataset = dataset.map(format_prompts, batched=True, num_proc=1)

# 5. Fast Fine-Tuning Setup
trainer = SFTTrainer(
    model=model,
    tokenizer=tokenizer,
    train_dataset=dataset,
    dataset_text_field="text",
    max_seq_length=MAX_SEQ_LENGTH,
    dataset_num_proc=1,
    packing=False,

    args=TrainingArguments(
        per_device_train_batch_size=8,      # Increased per-GPU batch size
        gradient_accumulation_steps=2,      # Effective Global Batch Size = 8 * 2 = 16
        warmup_steps=20,
        max_steps=400,                      # Reduced steps: model converges around 300–400 steps
        learning_rate=2e-4,
        fp16=not is_bfloat16_supported(),
        bf16=is_bfloat16_supported(),
        logging_steps=10,
        optim="adamw_8bit",
        weight_decay=0.01,
        lr_scheduler_type="cosine",
        seed=42,
        output_dir="outputs",
        save_strategy="no",
        report_to="tensorboard",
    ),
)

# Mask loss to doctor responses
trainer = train_on_responses_only(
    trainer,
    instruction_part="<|im_start|>user\n",
    response_part="<|im_start|>assistant\n",
    num_proc=1,
)

# 6. Train and Save Adapter
trainer.train()
model.save_pretrained(f"{OUTPUT_DIR}/final_adapter")
tokenizer.save_pretrained(f"{OUTPUT_DIR}/final_adapter")
print(f"\n🎉 Training complete! Adapter saved to {OUTPUT_DIR}/final_adapter")
```

```python
import os
import pandas as pd
import torch
from tqdm import tqdm
from unsloth import FastLanguageModel

# 1. Paths & Configuration
LOCAL_ADAPTER = "/kaggle/working/bn-doctor-qwen3b-lora/final_adapter"
KAGGLE_ADAPTER = "/kaggle/input/datasets/towshintushi/new-adaptor"

ADAPTER_PATH = LOCAL_ADAPTER if os.path.exists(LOCAL_ADAPTER) else KAGGLE_ADAPTER

SYSTEM_PROMPT = (
    "আপনি একজন অভিজ্ঞ ও সহানুভূতিশীল চিকিৎসক। রোগীর বর্ণনা পড়ে "
    "স্পষ্ট, নির্ভুল এবং সহজবোধ্য ভাষায় চিকিৎসা পরামর্শ দিন।"
)

# 2. Load Model & Prepare for Fast Inference
try:
    FastLanguageModel.for_inference(model)
    print("Using model already loaded in memory.")
except NameError:
    print(f"Loading model from {ADAPTER_PATH}...")
    model, tokenizer = FastLanguageModel.from_pretrained(
        model_name=ADAPTER_PATH,
        max_seq_length=1024,
        load_in_4bit=True,
    )
    FastLanguageModel.for_inference(model)

tokenizer.padding_side = "left"
if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token

eos_token_id = tokenizer.convert_tokens_to_ids("<|im_end|>")
if eos_token_id is None:
    eos_token_id = tokenizer.eos_token_id


def clean_doctor_response(text):
    if not isinstance(text, str) or not text:
        return ""
    stop_markers = [
        "<|im_end|>",
        "<|im_start|>",
        "<|endoftext|>",
        "user\n",
        "assistant\n",
    ]
    for marker in stop_markers:
        if marker in text:
            text = text.split(marker)[0]
    return text.strip()


# 3. Load Cleaned Test Data
test_df = df_base_test
preds = []
BATCH_SIZE = 16

print(
    f"Generating predictions for {len(test_df)} test queries in batches of {BATCH_SIZE}..."
)

# 4. Batch Inference Loop
for i in tqdm(range(0, len(test_df), BATCH_SIZE)):
    batch_df = test_df.iloc[i : i + BATCH_SIZE]
    batch_prompts = batch_df["input"].tolist()

    texts = [
        tokenizer.apply_chat_template(
            [
                {"role": "system", "content": SYSTEM_PROMPT},
                {"role": "user", "content": p},
            ],
            tokenize=False,
            add_generation_prompt=True,
        )
        for p in batch_prompts
    ]

    inputs = tokenizer(
        texts,
        return_tensors="pt",
        padding=True,
        truncation=True,
        max_length=1024,
    ).to("cuda")

    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            max_new_tokens=350,
            repetition_penalty=1.05,  # Lowered from 1.25 to prevent character distortion
            do_sample=True,
            temperature=0.2,          # Low temperature ensures high precision
            top_p=0.85,
            eos_token_id=eos_token_id,
            pad_token_id=tokenizer.pad_token_id,
            # no_repeat_ngram_size REMOVED to prevent breaking multi-byte Bengali unicode subwords
        )

    # 5. Extract & Clean Generations per sample
    for j in range(len(batch_prompts)):
        sample_input_len = inputs["input_ids"][j].shape[0]
        gen_tokens = outputs[j][sample_input_len:]
        raw_text = tokenizer.decode(gen_tokens, skip_special_tokens=True)
        preds.append(clean_doctor_response(raw_text))

# 6. Save submission.csv
submission = pd.DataFrame({"id": test_df["id"], "output": preds})

submission.to_csv("submission.csv", index=False)

print("\n🎉 SUCCESS! submission.csv created successfully.")
print(f"Total rows: {len(submission)}")
print(submission.head(3))
```
